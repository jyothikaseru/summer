##### Why can't a sentiment classifier perform translation?
A sentiment classifier is a many-to-one model. It takes an input sequence and produces a single output (e.g., positive or negative). Translation is a sequence-to-sequence task, where an input sequence must be converted into an output sequence of variable length.

##### What does the encoder do? 
The encoder reads the input sequence one token at a time using an LSTM. As it processes each token, it updates its hidden state. After reading the complete sequence, the final hidden state (and sometimes the cell state) forms the context vector, which summarizes the input.

##### 3.What does the decoder do?

The decoder takes the context vector as its initial information and generates the output sequence one token at a time until it reaches the end-of-sequence token.

##### 4. What is the context vector?
The context vector is the encoder's final representation of the input sequence. It summarizes the important information from the input sentence and passes it to the decoder.

##### 5. Why is it called a bottleneck?
The context vector usually has a fixed size (for example, 512 dimensions), no matter how long the sentence is

The encoder must compress all the information from the input sentence, whether it's 5 words or 100 words, into one fixed-size context vector. As sentences become longer, important information can be lost.

##### 6. Why do long sentences cause problems?
In long sentences, the encoder may not be able to preserve all important information in a single context vector, so details—especially from the beginning of the sentence—can be lost.

##### 7. What idea eventually solved this bottleneck?
The Attention Mechanism allows the decoder to focus on different parts of the input sequence instead of relying only on a single fixed context vector.
