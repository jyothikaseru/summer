```python
import torch
import torch.nn as nn
help(nn.RNN)
```

```python
rnn = nn. RNN(
    input_size=10,
    hidden_size=20,
    batch_first=True
)

x=torch.rand(4,5,10)

output, hidden = rnn(x)

print(output.shape)
print(hidden.shape)
```

```python
rnn = nn. RNN(
    input_size=10,
    hidden_size=20,
    num_layers=3,
    batch_first=True
)

x=torch.rand(4,5,10)

output, hidden = rnn(x)

print(output.shape)
print(hidden.shape)
```
