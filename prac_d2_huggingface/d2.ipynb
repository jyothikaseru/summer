#### tokenization


```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

text = "I love artificial intelligence"

tokens = tokenizer.tokenize(text)

print(tokens)
ids = tokenizer.encode(text)

print(ids)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

model = SentenceTransformer('all-MiniLM-L6-v2')
docs=["I love machine learning", "Artificial intelligence is fascinating", "I enjoy deep learning"]
embeddings = model.encode(docs)

sentence="are you interested in machine learning?"
sen_embedding = model.encode([sentence])
similarity_scores = cosine_similarity(sen_embedding, embeddings)
print(similarity_scores)
most_similar_doc_index = np.argmax(similarity_scores)
print("Most similar document:", docs[most_similar_doc_index])
```
